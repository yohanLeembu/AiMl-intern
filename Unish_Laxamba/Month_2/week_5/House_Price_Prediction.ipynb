# Data Cleaning

```python
# Preforming the further process using the copy of the orginal data for backup purpose
df = df_initial.copy()
```

```python
# Changing the data type of the pricing related columns
price_columns = df.iloc[:,-4:].columns.to_list()

for col in price_columns:
    df[col] = df[col].astype(float)
```

```python
df.info()
```

# Data Preprocessing

```python
# Encoding the categorical column (Location)
from sklearn.preprocessing import OneHotEncoder

location_encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoded_array = location_encoder.fit_transform(df[['location']])
encoded_df = pd.DataFrame(list(encoded_array), columns= location_encoder.get_feature_names_out())
encoded_df = encoded_df.astype(int)
df = pd.concat([df.drop(columns=['location']), encoded_df], axis=1)
```

```python
# The value of the total price is wrong i have corrected it
df = df.drop(columns=['total_price'])
df['total_price'] = df['land_cost'] + df['construction_cost'] + df['material_cost']
df.head()
```

# Final EDA

# Feature Engineering

After analyzing the correlation between the feature variables itself and between the feature variables and target variable, most of the
feature variables makes less correlation with the target variable so, that those variables are removed for creation of optimum predictive
model.

```python
df.head()
```

```python
# Creating the X and y variables (a.ka feature variables and target variable)
X = df.drop(columns=['land_cost', 'construction_cost', 'material_cost', 'rcc_structure', 'plumbing', 'electricity', 'windows','doors', 'cement_bags', 'total_price'])
y = df['total_price']
```

# Data Splitting

```python
# Splitting the data into tranning and testing data, also scaling the X test and train data
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

std_scale = StandardScaler()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

#Scaling the data
X_train_scaled = std_scale.fit_transform(X_train)
X_test_scaled = std_scale.transform(X_test)
```

# Model Creation

```python
# Model creation
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train_scaled, y_train)
```

# Model Evaluation

```python
# Creating the Testing and Trainning Prediction
Test_pred = model.predict(X_test_scaled)
Train_pred = model.predict(X_train_scaled)
```

```python
# Checking the model testing and tranning performance
from sklearn.metrics import r2_score
Train_performance = r2_score(y_train, Train_pred)
Test_performance = r2_score(y_test, Test_pred)

print(f'The overall testing performance is: {Test_performance}')
print(f'The overall tranning performance is: {Train_performance}')
```

AS we can see that the testing and tranning preformance is very good, so in conclusion the overall model will perfectly predict the required
house price as per the numbers of (floors, bedrooms, bathrooms) and the preferred location (chitwan, kathmandu, dhangadi, pokhara, syangja).

# Importing the library

```python
import numpy as np
```

# Array Creation

```python
data = np.array([1,2,3])
data1 = np.array([[1,2,3], [4,5,6]])

print(f'1D array: {data}')
print(' ')
print(f'2D array: \n{data1}')
print('\nCreated array using the array function.')
```

```python
print(f'{np.zeros((3,2))}')
print('\nCreated array using the zeros function.')
```

```python
print(f'{np.ones((3,2))}')
print('\nCreated array using the ones function.')
```

```python
print(f'{np.arange(0,11,1)}')
print('\nCreated array using the arange function.')
```

```python
print(f'{np.linspace(0,5,3)}')
print('\nCreated array using the linespace function.')
```

# Indexing in array

```python
print(f'First element of the 1D array is: {data[0]}')
print(f'Last element of the 1D array is: {data[-1]}')
print(f'Middle element of the 1D array is: {data[1:2][0]}')
```

```python
print(f'First element of the 2D array: {data1[0][0]}')
print(f'Last element of the 2D array: {data1[1][2]}')
```

# Broadcasting the Array

```python
print(f'Brodacasting the 1D array by adding {2} in each element inside the array {data}, the result is: {2 + data}')
```

```python
print(f'Brodacasting the 2D array by multiplying the each element inside the array \n{data1} by {2}, \nthe result is: \n{2 + data1}')
```

# Vectorized OPS

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = a + b 
print(f'Vectorized addition between the {a} and {b}, the result is: {c}')
```

```python
arr = np.array([1, 2, 3, 4])

result = arr * 2  # Output: [2, 4, 6, 8]
print(f'Vectorized multiplication (multiplies each element by 2 instantly, the result is: {result}')

squared_roots = np.sqrt(arr)
print(f'Vectorized math function, the result is: {squared_roots}')
```

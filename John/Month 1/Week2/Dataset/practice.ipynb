```python
import numpy as np
import pandas as pd
```

```python
#reading the csv file
poke = pd.read_csv('Pokemon.csv')
#asking for the top 10
poke.head(10)
```

```python
# True is treated as 1 and False as 0, so sum() counts missing values per column
poke.isnull().sum()

#replacing the missin value with 0
poke1 = poke.fillna(0)

#drops the whole row where NaN values are
poke1 = poke.dropna()
```

```python
#showing 10 pokemon who have type1 fire
fire_pokemon = poke[poke['Type1']=='Fire']

print(fire_pokemon.head(10))

#showing pokemon who has atk higher than 100
high_atk_poke = poke[poke['Attack'] > 100]

print(high_atk_poke.head(10))
```

```python
# Groups Pokemon by their Type1 category, then calculates the average Attack stat for each type
avg_atk = poke.groupby('Type1')['Attack'].mean()
 
print(avg_atk)
```

```python
#Pivot 
#pivot table is matching the index(type1) with column(type2) and 
#then finding the pokemon that matches it and then sending their atk values for calculation
pivot = pd.pivot_table(
    poke,
    values= 'Attack',
    index='Type1',
    columns='Type2',
    aggfunc='mean'
)

print(pivot)
```

```python
#created a dataset to merge with our csv dataset
regions = pd.DataFrame({
    "Generation":[1,2,3,4,5,6],
    "Region":[
        "Kanto",
        "Johto",
        "Hoenn",
        "Sinnoh",
        "Unova",
        "Kalos"
    ]
})
print(regions)

#merging both dataset by using generation as a common ground
merged = pd.merge(
        poke,
        regions,
        on = 'Generation'
)
print(merged)
```

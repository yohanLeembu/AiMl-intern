```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])
arr2 = np.ones(5)
arr3 = np.arange(10)

print(arr)
print(type(arr))
print(arr + 10) #can work with operators unlike a list
print(arr[:4])
print(arr[3:4])
print(arr[2:])
print(arr2)
print(arr3 *2)
```

```python
#same size array can be added too
a = np.array([1, 2, 3])

b = np.array([10, 20, 30])

print(a+b)

print(a.mean())
print(a.max())
print(a.min())
print(a.sum())
```

```python
import numpy as np
import pandas as pd

print("Libraries imported successfully!")


data = {
    "ID": [101, 102, 103, 104, 105, 106, 107, 108, 108],
    
    "Name": [
        "Alice",
        "Bob",
        "Charlie",
        "David",
        "Emma",
        "Frank",
        "Grace",
        "Henry",
        "Henry"
    ],
    
    "Age": [
        25,
        35,
        np.nan,
        42,
        31,
        27,
        39,
        45,
        45
    ],
    
    "Gender": [
        "Female",
        "Male",
        "Male",
        "Male",
        "Female",
        "Male",
        "Female",
        "Male",
        "Male"
    ],
    
    "Department": [
        "HR",
        "IT",
        "IT",
        "Finance",
        "HR",
        "Finance",
        "IT",
        "HR",
        "HR"
    ],
    
    "Salary": [
        50000,
        65000,
        58000,
        np.nan,
        61000,
        55000,
        69000,
        75000,
        75000
    ]
}

df = pd.DataFrame(data)

print("\n========== RAW DATASET ==========")
print(df)


df.to_csv("raw_employee_data.csv", index=False)

print("\nRaw dataset saved as 'raw_employee_data.csv'")

df = pd.read_csv("raw_employee_data.csv")

print("\n========== DATASET LOADED ==========")
print(df)

```

```python
#EXPLORE THE DATASET

print("\n========== FIRST FIVE ROWS ==========")
print(df.head())


print("\n========== DATASET INFORMATION ==========")
df.info()


print("\n========== STATISTICAL SUMMARY ==========")
print(df.describe())
```

```python
# STEP 6: CHECK DATASET SIZE


print("\n========== DATASET SHAPE ==========")

rows, columns = df.shape

print("Number of rows:", rows)
print("Number of columns:", columns)
```

```python
# CHECK MISSING VALUES

print("\n========== MISSING VALUES ==========")

print(df.isnull().sum())
```

```python


#HANDLE MISSING VALUES


# Fill missing Age with the average age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill missing Salary with the average salary
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("\n========== MISSING VALUES AFTER CLEANING ==========")

print(df.isnull().sum())

```

```python

# CHECK DUPLICATE ROWS
print("\n========== DUPLICATE ROWS ==========")

print("Number of duplicate rows:", df.duplicated().sum())

```

```python


# REMOVE DUPLICATES


df.drop_duplicates(inplace=True)

print("\nDuplicate rows removed successfully!")

print("Number of rows after removing duplicates:", len(df))
```

```python



# RENAME COLUMNS
df.rename(columns={
    "ID": "Employee_ID",
    "Name": "Employee_Name",
    "Age": "Employee_Age",
    "Gender": "Employee_Gender",
    "Salary": "Employee_Salary"
}, inplace=True)

print("\n========== RENAMED COLUMNS ==========")

print(df.columns)

```

```python

# STEP 12: FILTER DATA


print("\n========== EMPLOYEES OLDER THAN 30 ==========")

older_employees = df[df["Employee_Age"] > 30]

print(older_employees)

```

```python


# FILTER HIGH SALARY EMPLOYEES


print("\n========== EMPLOYEES WITH SALARY ABOVE 60000 ==========")

high_salary = df[df["Employee_Salary"] > 60000]

print(high_salary)
```

```python
# SORT DATA

print("\n========== SORTED BY SALARY ==========")

sorted_salary = df.sort_values(
    by="Employee_Salary",
    ascending=False
)

print(sorted_salary)
```

```python



# SELECT SPECIFIC COLUMNS


print("\n========== EMPLOYEE NAME AND SALARY ==========")

selected_columns = df[
    ["Employee_Name", "Department", "Employee_Salary"]
]

print(selected_columns)
```

```python



# GROUPBY - AVERAGE SALARY


print("\n========== AVERAGE SALARY BY DEPARTMENT ==========")

average_salary = df.groupby(
    "Department"
)["Employee_Salary"].mean()

print(average_salary)
```

```python

# STEP 17: GROUPBY - AVERAGE AGE

print("\n========== AVERAGE AGE BY DEPARTMENT ==========")

average_age = df.groupby(
    "Department"
)["Employee_Age"].mean()

print(average_age)
```

```python



# COUNT EMPLOYEES BY DEPARTMENT


print("\n========== EMPLOYEE COUNT BY DEPARTMENT ==========")

employee_count = df.groupby(
    "Department"
)["Employee_ID"].count()

print(employee_count)

```

```python


# CREATE PIVOT TABLE


print("\n========== PIVOT TABLE ==========")

pivot = df.pivot_table(
    values="Employee_Salary",
    index="Department",
    columns="Employee_Gender",
    aggfunc="mean"
)

print(pivot)

```

```python


# CREATE SECOND PIVOT TABLE


print("\n========== EMPLOYEE COUNT PIVOT TABLE ==========")

count_pivot = df.pivot_table(
    values="Employee_ID",
    index="Department",
    columns="Employee_Gender",
    aggfunc="count",
    fill_value=0
)

print(count_pivot)

```

```python


# FINAL DATASET


print("\n========== FINAL CLEANED DATASET ==========")

print(df)

```

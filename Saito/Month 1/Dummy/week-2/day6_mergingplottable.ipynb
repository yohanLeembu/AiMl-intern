```python

import pandas as pd
employees = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Department": ["HR", "IT", "IT", "Finance", "HR"]
}

df1 = pd.DataFrame(employees)

print("========== EMPLOYEE DATA ==========")
print(df1)

salary = {
    "ID": [101, 102, 103, 104, 105],
    "Salary": [50000, 65000, 58000, 72000, 61000]
}

df2 = pd.DataFrame(salary)

print("\n========== SALARY DATA ==========")
print(df2)
```

```python
#merging 2 data

merged = pd.merge(df1, df2, on="ID")

print("\n========== MERGED DATA ==========")
print(merged)

```

```python

#join 2 data frames

# Set ID as the index before joining
df1_indexed = df1.set_index("ID")
df2_indexed = df2.set_index("ID")

joined = df1_indexed.join(df2_indexed)

print("\n========== JOINED DATA ==========")
print(joined)

```

```python

#concatenating 2 data frames

new_employees = {
    "ID": [106, 107],
    "Name": ["Frank", "Grace"],
    "Department": ["Finance", "IT"]
}

df3 = pd.DataFrame(new_employees)

concatenated = pd.concat([df1, df3], ignore_index=True)

print("\n========== CONCATENATED DATA ==========")
print(concatenated)
```

```python
# Create Dataset for Pivot Table


sales_data = {
    "Employee": [
        "Alice", "Alice",
        "Bob", "Bob",
        "Charlie", "Charlie",
        "David", "David",
        "Emma", "Emma"
    ],
    
    "Department": [
        "HR", "HR",
        "IT", "IT",
        "IT", "IT",
        "Finance", "Finance",
        "HR", "HR"
    ],
    
    "Month": [
        "January", "February",
        "January", "February",
        "January", "February",
        "January", "February",
        "January", "February"
    ],
    
    "Sales": [
        5000, 6000,
        7000, 7500,
        6500, 8000,
        9000, 8500,
        5500, 6200
    ]
}

sales_df = pd.DataFrame(sales_data)

print("\n========== SALES DATA ==========")
print(sales_df)

```

```python


#Create Pivot Table


pivot = sales_df.pivot_table(
    values="Sales",
    index="Department",
    columns="Month",
    aggfunc="mean"
)

print("\n========== PIVOT TABLE ==========")
print(pivot)

```

```python


#Pivot Table by Employee


employee_pivot = sales_df.pivot_table(
    values="Sales",
    index="Employee",
    columns="Month",
    aggfunc="sum"
)

print("\n========== EMPLOYEE SALES PIVOT TABLE ==========")
print(employee_pivot)
```

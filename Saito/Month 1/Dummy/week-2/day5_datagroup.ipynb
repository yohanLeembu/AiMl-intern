```python
import pandas as pd
#creating new data set using dictionary
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma", "Frank", "Grace", "Henry"],
    "Age": [25, 35, 29, 42, 31, 27, 39, 45],
    "Gender": ["Female", "Male", "Male", "Male", "Female", "Male", "Female", "Male"],
    "Department": ["HR", "IT", "IT", "Finance", "HR", "Finance", "IT", "HR"],
    "Salary": [50000, 65000, 58000, 72000, 61000, 55000, 69000, 75000]
}

df = pd.DataFrame(data)

print("========== Original Dataset ==========")
print(df)

```

```python
#filtered data for employees older than 30

print("\n========== Employees Older Than 30 ==========")

filtered = df[df["Age"] > 30]

print(filtered)

```

```python
#short by salary in ascending order

print("\n========== Sort by Salary (Ascending) ==========")

sorted_salary = df.sort_values(by="Salary")

print(sorted_salary)

print("\n========== Sort by Age (Descending) ==========")

sorted_age = df.sort_values(by="Age", ascending=False)

print(sorted_age)

```

```python
#select specific columns

print("\n========== Name and Salary Columns ==========")

selected_columns = df[["Name", "Salary"]]

print(selected_columns)

```

```python
#group by

print("\n========== Average Salary by Department ==========")

avg_salary = df.groupby("Department")["Salary"].mean()

print(avg_salary)

```

```python
#caluclate average age by gender

print("\n========== Average Age by Gender ==========")

avg_age = df.groupby("Gender")["Age"].mean()

print(avg_age)

```

```python
#count employees in each department

print("\n========== Employee Count by Department ==========")

department_count = df.groupby("Department").count()

print(department_count)

```

```python
#count employees only

print("\n========== Number of Employees Per Department ==========")

employee_count = df.groupby("Department")["Employee_ID"].count()

print(employee_count)

```

```python

#max salary by department

print("\n========== Highest Salary by Department ==========")

max_salary = df.groupby("Department")["Salary"].max()

print(max_salary)
```

```python
#minimu salary by department

print("\n========== Lowest Salary by Department ==========")

min_salary = df.groupby("Department")["Salary"].min()

print(min_salary)
```

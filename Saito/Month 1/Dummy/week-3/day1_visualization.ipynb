```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")



# CREATE A NEW EMPLOYEE DATASET


data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "Name": [
        "Alice",
        "Bob",
        "Charlie",
        "David",
        "Emma",
        "Frank",
        "Grace",
        "Henry",
        "Ivy",
        "Jack"
    ],

    "Age": [
        24, 29, 35, 42, 31,
        26, 38, 45, 28, 33
    ],

    "Gender": [
        "Female",
        "Male",
        "Male",
        "Male",
        "Female",
        "Male",
        "Female",
        "Male",
        "Female",
        "Male"
    ],

    "Department": [
        "HR",
        "IT",
        "IT",
        "Finance",
        "HR",
        "Finance",
        "IT",
        "HR",
        "Finance",
        "IT"
    ],

    "Salary": [
        45000,
        55000,
        65000,
        75000,
        52000,
        58000,
        70000,
        80000,
        60000,
        68000
    ]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")



#  DISPLAY COMPLETE DATASET


print("\n========== DATASET ==========")

print(df)
```

```python
#  DISPLAY FIRST 5 ROWS


print("\n========== FIRST 5 ROWS ==========")

print(df.head())



# DATASET INFORMATION


print("\n========== DATASET INFORMATION ==========")

df.info()



# DATASET SIZE


print("\n========== DATASET SIZE ==========")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])



#  STATISTICAL SUMMARY


print("\n========== STATISTICAL SUMMARY ==========")

print(df.describe())



# CHECK DEPARTMENTS


print("\n========== DEPARTMENTS ==========")

print(df["Department"].unique())



# COUNT EMPLOYEES BY DEPARTMENT


department_counts = df["Department"].value_counts()

print("\n========== EMPLOYEES BY DEPARTMENT ==========")

print(department_counts)

```

```python


# DEPARTMENT BAR CHART


plt.figure(figsize=(8, 5))

plt.bar(
    department_counts.index,
    department_counts.values
)

plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()



# COUNT EMPLOYEES BY GENDER


gender_counts = df["Gender"].value_counts()

print("\n========== EMPLOYEES BY GENDER ==========")

print(gender_counts)



#  GENDER BAR CHART


plt.figure(figsize=(8, 5))

plt.bar(
    gender_counts.index,
    gender_counts.values
)

plt.title("Number of Employees by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")

plt.show()



#  EMPLOYEE SALARY CHART


plt.figure(figsize=(10, 5))

plt.bar(
    df["Name"],
    df["Salary"]
)

plt.title("Employee Salary")
plt.xlabel("Employee")
plt.ylabel("Salary")

plt.xticks(rotation=45)

plt.show()



# EMPLOYEE AGE CHART


plt.figure(figsize=(10, 5))

plt.bar(
    df["Name"],
    df["Age"]
)

plt.title("Employee Age")
plt.xlabel("Employee")
plt.ylabel("Age")

plt.xticks(rotation=45)

plt.show()



# SAVE DATASET
df.to_csv("week3_employee_data.csv", index=False)
```

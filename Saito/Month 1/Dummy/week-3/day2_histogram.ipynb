```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")


df = pd.read_csv("week3_employee_data.csv")

print("Dataset loaded successfully!")

print("\n========== DATASET ==========")

print(df)

print("\n========== DATASET INFORMATION ==========")

df.info()


print("\n========== EMPLOYEE AGE ==========")

print(df["Age"])



print("\n========== EMPLOYEE SALARY ==========")

print(df["Salary"])


print("\n========== AGE STATISTICS ==========")

print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())
print("Average Age:", df["Age"].mean())

print("\n========== SALARY STATISTICS ==========")

print("Minimum Salary:", df["Salary"].min())
print("Maximum Salary:", df["Salary"].max())
print("Average Salary:", df["Salary"].mean())


```

```python

# CREATE AGE HISTOGRAM USING MATPLOTLIB


plt.figure(figsize=(8, 5))

plt.hist(
    df["Age"],
    bins=5
)

plt.title("Distribution of Employee Age")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.show()



# CREATE SALARY HISTOGRAM USING MATPLOTLIB


plt.figure(figsize=(8, 5))

plt.hist(
    df["Salary"],
    bins=5
)

plt.title("Distribution of Employee Salary")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()



#  CREATE AGE HISTOGRAM USING SEABORN


plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=5,
    kde=True
)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()



#  CREATE SALARY HISTOGRAM USING SEABORN


plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Salary",
    bins=5,
    kde=True
)

plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()



# DISPLAY AGE DISTRIBUTION COUNTS


print("\n========== AGE DISTRIBUTION ==========")

age_bins = pd.cut(df["Age"], bins=5)

print(age_bins.value_counts().sort_index())



#  DISPLAY SALARY DISTRIBUTION COUNTS


print("\n========== SALARY DISTRIBUTION ==========")

salary_bins = pd.cut(df["Salary"], bins=5)

print(salary_bins.value_counts().sort_index())



# FIND THE OLDEST AND YOUNGEST EMPLOYEE


youngest_employee = df.loc[df["Age"].idxmin()]
oldest_employee = df.loc[df["Age"].idxmax()]

print("\n========== AGE ANALYSIS ==========")

print(
    "Youngest Employee:",
    youngest_employee["Name"],
    "- Age:",
    youngest_employee["Age"]
)

print(
    "Oldest Employee:",
    oldest_employee["Name"],
    "- Age:",
    oldest_employee["Age"]
)



#  FIND LOWEST AND HIGHEST SALARY


lowest_salary = df.loc[df["Salary"].idxmin()]
highest_salary = df.loc[df["Salary"].idxmax()]

print("\n========== SALARY ANALYSIS ==========")

print(
    "Lowest Salary:",
    lowest_salary["Name"],
    "- Salary:",
    lowest_salary["Salary"]
)

print(
    "Highest Salary:",
    highest_salary["Name"],
    "- Salary:",
    highest_salary["Salary"]
)

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("week3_employee_data.csv")

print("Dataset loaded successfully!")



print("\n========== DATASET ==========")

print(df)




print("\n========== DATASET INFORMATION ==========")

df.info()
```

```python
# SELECT NUMERICAL COLUMNS
numeric_data = df[
    ["Employee_ID", "Age", "Salary"]
]

print("\n========== NUMERICAL DATA ==========")

print(numeric_data)

```

```python


#  CALCULATE CORRELATION MATRIX


correlation_matrix = numeric_data.corr()

print("\n========== CORRELATION MATRIX ==========")

print(correlation_matrix)

```

```python


#  CORRELATION BETWEEN AGE AND SALARY


age_salary_correlation = df["Age"].corr(df["Salary"])

print("\n========== AGE AND SALARY CORRELATION ==========")

print(
    "Correlation between Age and Salary:",
    age_salary_correlation
)

```

```python


# INTERPRET AGE AND SALARY CORRELATION


print("\n========== CORRELATION INTERPRETATION ==========")

if age_salary_correlation >= 0.7:

    print("Strong positive correlation.")

elif age_salary_correlation >= 0.3:

    print("Moderate positive correlation.")

elif age_salary_correlation > -0.3:

    print("Weak or no correlation.")

elif age_salary_correlation > -0.7:

    print("Moderate negative correlation.")

else:

    print("Strong negative correlation.")


```

```python

# CREATE BASIC HEATMAP


plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()

```

```python

# CREATE DETAILED HEATMAP


plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Employee Data Correlation Heatmap")

plt.show()

```

```python


# CREATE HEATMAP WITH CENTER AT ZERO


plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Heatmap Centered at Zero")

plt.show()


```

```python

# FIND STRONGEST CORRELATION


print("\n========== CORRELATION VALUES ==========")

# Display all correlation values
print(correlation_matrix)



#  FIND AGE AND SALARY CORRELATION


age_salary_correlation = correlation_matrix.loc["Age", "Salary"]

print("\n========== AGE VS SALARY ==========")

print("Age and Salary Correlation:", age_salary_correlation)



#  INTERPRET CORRELATION


print("\n========== INTERPRETATION ==========")

if age_salary_correlation >= 0.7:
    print("There is a strong positive relationship between Age and Salary.")

elif age_salary_correlation >= 0.3:
    print("There is a moderate positive relationship between Age and Salary.")

elif age_salary_correlation > -0.3:
    print("There is a weak or no relationship between Age and Salary.")

elif age_salary_correlation > -0.7:
    print("There is a moderate negative relationship between Age and Salary.")

else:
    print("There is a strong negative relationship between Age and Salary.")



#  FIND HIGHEST SALARY


highest_salary = df.loc[df["Salary"].idxmax()]

print("\n========== HIGHEST SALARY ==========")

print("Employee:", highest_salary["Name"])
print("Age:", highest_salary["Age"])
print("Salary:", highest_salary["Salary"])



#  FIND LOWEST SALARY


lowest_salary = df.loc[df["Salary"].idxmin()]

print("\n========== LOWEST SALARY ==========")

print("Employee:", lowest_salary["Name"])
print("Age:", lowest_salary["Age"])
print("Salary:", lowest_salary["Salary"])



# AGE STATISTICS


print("\n========== AGE STATISTICS ==========")

print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())
print("Average Age:", df["Age"].mean())
print("Median Age:", df["Age"].median())



#  SALARY STATISTICS


print("\n========== SALARY STATISTICS ==========")

print("Minimum Salary:", df["Salary"].min())
print("Maximum Salary:", df["Salary"].max())
print("Average Salary:", df["Salary"].mean())
print("Median Salary:", df["Salary"].median())

```

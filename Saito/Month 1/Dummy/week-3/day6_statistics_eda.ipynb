```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("week3_employee_data.csv")

print("\nDataset loaded successfully!")


print("\n========== DATASET ==========")

print(df)



```

```python

print("\n========== DATASET SHAPE ==========")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

```

```python

print("\n========== COLUMN NAMES ==========")

print(df.columns.tolist())


```

```python

#  CHECK DATA TYPES


print("\n========== DATA TYPES ==========")

print(df.dtypes)

```

```python

# CHECK MISSING VALUES


print("\n========== MISSING VALUES ==========")

print(df.isnull().sum())


```

```python

# CHECK DUPLICATE ROWS


print("\n========== DUPLICATE ROWS ==========")

print("Number of duplicate rows:", df.duplicated().sum())

```

```python

#  DESCRIPTIVE STATISTICS


print("\n========== DESCRIPTIVE STATISTICS ==========")

print(df.describe())


```

```python

# AGE STATISTICS


print("\n========== AGE STATISTICS ==========")

age_mean = df["Age"].mean()
age_median = df["Age"].median()
age_mode = df["Age"].mode()[0]
age_min = df["Age"].min()
age_max = df["Age"].max()
age_std = df["Age"].std()

print("Mean Age:", age_mean)
print("Median Age:", age_median)
print("Mode Age:", age_mode)
print("Minimum Age:", age_min)
print("Maximum Age:", age_max)
print("Standard Deviation:", age_std)


```

```python

#  SALARY STATISTICS


print("\n========== SALARY STATISTICS ==========")

salary_mean = df["Salary"].mean()
salary_median = df["Salary"].median()
salary_mode = df["Salary"].mode()[0]
salary_min = df["Salary"].min()
salary_max = df["Salary"].max()
salary_std = df["Salary"].std()

print("Mean Salary:", salary_mean)
print("Median Salary:", salary_median)
print("Mode Salary:", salary_mode)
print("Minimum Salary:", salary_min)
print("Maximum Salary:", salary_max)
print("Standard Deviation:", salary_std)


```

```python

#  CALCULATE SALARY RANGE


salary_range = salary_max - salary_min

print("\n========== SALARY RANGE ==========")

print("Salary Range:", salary_range)

```

```python

#  CALCULATE AGE RANGE


age_range = age_max - age_min

print("\n========== AGE RANGE ==========")

print("Age Range:", age_range)


```

```python

#FIND OLDEST EMPLOYEE


oldest_employee = df.loc[df["Age"].idxmax()]

print("\n========== OLDEST EMPLOYEE ==========")

print("Name:", oldest_employee["Name"])
print("Age:", oldest_employee["Age"])
print("Salary:", oldest_employee["Salary"])

```

```python

#  FIND YOUNGEST EMPLOYEE


youngest_employee = df.loc[df["Age"].idxmin()]

print("\n========== YOUNGEST EMPLOYEE ==========")

print("Name:", youngest_employee["Name"])
print("Age:", youngest_employee["Age"])
print("Salary:", youngest_employee["Salary"])

```

```python


#  FIND HIGHEST PAID EMPLOYEE


highest_paid = df.loc[df["Salary"].idxmax()]

print("\n========== HIGHEST PAID EMPLOYEE ==========")

print("Name:", highest_paid["Name"])
print("Age:", highest_paid["Age"])
print("Salary:", highest_paid["Salary"])



#  FIND LOWEST PAID EMPLOYEE


lowest_paid = df.loc[df["Salary"].idxmin()]

print("\n========== LOWEST PAID EMPLOYEE ==========")

print("Name:", lowest_paid["Name"])
print("Age:", lowest_paid["Age"])
print("Salary:", lowest_paid["Salary"])



# AVERAGE SALARY BY DEPARTMENT


if "Department" in df.columns:

    department_salary = df.groupby(
        "Department"
    )["Salary"].mean()

    print("\n========== AVERAGE SALARY BY DEPARTMENT ==========")

    print(department_salary)

```

```python

#  NUMBER OF EMPLOYEES BY DEPARTMENT


if "Department" in df.columns:

    department_count = df["Department"].value_counts()

    print("\n========== EMPLOYEES BY DEPARTMENT ==========")

    print(department_count)


# BAR CHART OF EMPLOYEES BY DEPARTMENT


if "Department" in df.columns:

    plt.figure(figsize=(8, 5))

    df["Department"].value_counts().plot(
        kind="bar"
    )

    plt.title("Number of Employees by Department")
    plt.xlabel("Department")
    plt.ylabel("Number of Employees")

    plt.xticks(rotation=0)

    plt.show()



# AGE DISTRIBUTION


plt.figure(figsize=(8, 5))

plt.hist(
    df["Age"],
    bins=5
)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.show()



# SALARY DISTRIBUTION


plt.figure(figsize=(8, 5))

plt.hist(
    df["Salary"],
    bins=5
)

plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()


#  AGE VS SALARY


plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Salary"
)

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()



#  BOX PLOT OF SALARY


plt.figure(figsize=(8, 5))

sns.boxplot(
    y=df["Salary"]
)

plt.title("Salary Distribution and Outliers")
plt.ylabel("Salary")

plt.show()



# STEP 25 - CORRELATION MATRIX


numeric_columns = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_columns.corr()

print("\n========== CORRELATION MATRIX ==========")

print(correlation_matrix)



#  CORRELATION HEATMAP


plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Employee Data Correlation Heatmap")

plt.show()



# SORT EMPLOYEES BY SALARY


print("\n========== EMPLOYEES SORTED BY SALARY ==========")

salary_sorted = df.sort_values(
    by="Salary",
    ascending=False
)

print(
    salary_sorted[
        ["Name", "Age", "Salary"]
    ]
)



# EMPLOYEES ABOVE AVERAGE SALARY


print("\n========== ABOVE AVERAGE SALARY ==========")

above_average = df[
    df["Salary"] > salary_mean
]

print(
    above_average[
        ["Name", "Age", "Salary"]
    ]
)



# EMPLOYEES BELOW AVERAGE SALARY


print("\n========== BELOW AVERAGE SALARY ==========")

below_average = df[
    df["Salary"] < salary_mean
]

print(
    below_average[
        ["Name", "Age", "Salary"]
    ]
)



# FINAL EDA SUMMARY
print("\nDataset Summary:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nAge:")
print("Mean:", age_mean)
print("Median:", age_median)
print("Minimum:", age_min)
print("Maximum:", age_max)
print("Standard Deviation:", age_std)

print("\nSalary:")
print("Mean:", salary_mean)
print("Median:", salary_median)
print("Minimum:", salary_min)
print("Maximum:", salary_max)
print("Standard Deviation:", salary_std)

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

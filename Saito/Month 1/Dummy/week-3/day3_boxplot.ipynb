```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")


df = pd.read_csv("week3_employee_data.csv")

print("Dataset loaded successfully!")


print("\n========== DATASET ==========")

print(df)


print("\n========== DATASET INFORMATION ==========")

df.info()


```

```python

# CHECK AGE DATA


print("\n========== AGE DATA ==========")

print(df["Age"])


# CHECK SALARY DATA


print("\n========== SALARY DATA ==========")

print(df["Salary"])


```

```python

#  CREATE AGE BOXPLOT


plt.figure(figsize=(8, 5))

plt.boxplot(df["Age"])

plt.title("Box Plot of Employee Age")
plt.ylabel("Age")

plt.show()


```

```python

# CREATE SALARY BOXPLOT


plt.figure(figsize=(8, 5))

plt.boxplot(df["Salary"])

plt.title("Box Plot of Employee Salary")
plt.ylabel("Salary")

plt.show()


```

```python

# CREATE BOXPLOTS USING SEABORN


plt.figure(figsize=(8, 5))

sns.boxplot(y=df["Age"])

plt.title("Employee Age Box Plot")
plt.ylabel("Age")

plt.show()


```

```python

# SALARY BOXPLOT USING SEABORN


plt.figure(figsize=(8, 5))

sns.boxplot(y=df["Salary"])

plt.title("Employee Salary Box Plot")
plt.ylabel("Salary")

plt.show()

```

```python


#  CALCULATE AGE QUARTILES


Q1_age = df["Age"].quantile(0.25)
Q2_age = df["Age"].quantile(0.50)
Q3_age = df["Age"].quantile(0.75)

IQR_age = Q3_age - Q1_age

print("\n========== AGE QUARTILES ==========")

print("Q1 (25%):", Q1_age)
print("Median (50%):", Q2_age)
print("Q3 (75%):", Q3_age)
print("IQR:", IQR_age)

```

```python

#  CALCULATE AGE OUTLIER LIMITS


lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

print("\n========== AGE OUTLIER LIMITS ==========")

print("Lower Limit:", lower_age)
print("Upper Limit:", upper_age)

```

```python


#  FIND AGE OUTLIERS


age_outliers = df[
    (df["Age"] < lower_age) |
    (df["Age"] > upper_age)
]

print("\n========== AGE OUTLIERS ==========")

if age_outliers.empty:
    print("No age outliers found.")
else:
    print(age_outliers)

```

```python


#  CALCULATE SALARY QUARTILES


Q1_salary = df["Salary"].quantile(0.25)
Q2_salary = df["Salary"].quantile(0.50)
Q3_salary = df["Salary"].quantile(0.75)

IQR_salary = Q3_salary - Q1_salary

print("\n========== SALARY QUARTILES ==========")

print("Q1 (25%):", Q1_salary)
print("Median (50%):", Q2_salary)
print("Q3 (75%):", Q3_salary)
print("IQR:", IQR_salary)


```

```python

#  CALCULATE SALARY OUTLIER LIMITS


lower_salary = Q1_salary - 1.5 * IQR_salary
upper_salary = Q3_salary + 1.5 * IQR_salary

print("\n========== SALARY OUTLIER LIMITS ==========")

print("Lower Limit:", lower_salary)
print("Upper Limit:", upper_salary)


```

```python

#  FIND SALARY OUTLIERS


salary_outliers = df[
    (df["Salary"] < lower_salary) |
    (df["Salary"] > upper_salary)
]

print("\n========== SALARY OUTLIERS ==========")

if salary_outliers.empty:
    print("No salary outliers found.")
else:
    print(salary_outliers)

```

```python


# CREATE A COMBINED BOXPLOT


plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df[["Age", "Salary"]]
)

plt.title("Box Plot of Age and Salary")

plt.show()
```

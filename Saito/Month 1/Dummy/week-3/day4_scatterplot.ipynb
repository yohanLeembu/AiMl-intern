```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")

df = pd.read_csv("week3_employee_data.csv")

print("Dataset loaded successfully!")



print("\n========== DATASET ==========")

print(df)


print("\n========== DATASET INFORMATION ==========")

df.info()

```

```python
#  DISPLAY AGE AND SALARY
print("\n========== AGE AND SALARY ==========")

print(df[["Name", "Age", "Salary"]])

```

```python
# CREATE SCATTER PLOT
# AGE VS SALARY
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Age"],
    df["Salary"]
)

plt.title("Employee Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()

```

```python

#  CREATE SCATTER PLOT WITH SEABORN
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Salary"
)

plt.title("Relationship Between Employee Age and Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()


```

```python

# ADD EMPLOYEE NAMES TO SCATTER PLOT


plt.figure(figsize=(10, 6))

plt.scatter(
    df["Age"],
    df["Salary"]
)

for i in range(len(df)):
    plt.annotate(
        df["Name"][i],
        (df["Age"][i], df["Salary"][i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Employee Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()


```

```python

# CALCULATE CORRELATION


correlation = df["Age"].corr(df["Salary"])

print("\n========== CORRELATION ==========")

print("Correlation between Age and Salary:", correlation)


```

```python

# INTERPRET CORRELATION


print("\n========== CORRELATION INTERPRETATION ==========")

if correlation > 0.7:
    print("There is a strong positive relationship between Age and Salary.")

elif correlation > 0.3:
    print("There is a moderate positive relationship between Age and Salary.")

elif correlation > -0.3:
    print("There is a weak or no relationship between Age and Salary.")

elif correlation > -0.7:
    print("There is a moderate negative relationship between Age and Salary.")

else:
    print("There is a strong negative relationship between Age and Salary.")


```

```python

# CREATE REGRESSION PLOT


plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Age",
    y="Salary"
)

plt.title("Age vs Salary with Trend Line")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()


```

```python

#  CHECK AGE VS EMPLOYEE ID


plt.figure(figsize=(8, 5))

plt.scatter(
    df["Employee_ID"],
    df["Age"]
)

plt.title("Employee ID vs Age")
plt.xlabel("Employee ID")
plt.ylabel("Age")

plt.show()

```

```python


# CHECK SALARY VS EMPLOYEE ID


plt.figure(figsize=(8, 5))

plt.scatter(
    df["Employee_ID"],
    df["Salary"]
)

plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary")

plt.show()


```

```python

#  DISPLAY CORRELATION MATRIX


print("\n========== CORRELATION MATRIX ==========")

correlation_matrix = df[
    ["Employee_ID", "Age", "Salary"]
].corr()

print(correlation_matrix)

```

```python


# CREATE CORRELATION HEATMAP


plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()


```

```python

# FIND HIGHEST SALARY


highest_salary = df.loc[df["Salary"].idxmax()]

print("\n========== HIGHEST SALARY ==========")

print("Employee:", highest_salary["Name"])
print("Age:", highest_salary["Age"])
print("Salary:", highest_salary["Salary"])

```

```python


# FIND LOWEST SALARY


lowest_salary = df.loc[df["Salary"].idxmin()]

print("\n========== LOWEST SALARY ==========")

print("Employee:", lowest_salary["Name"])
print("Age:", lowest_salary["Age"])
print("Salary:", lowest_salary["Salary"])
```

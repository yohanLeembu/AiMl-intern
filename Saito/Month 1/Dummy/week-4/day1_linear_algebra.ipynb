```python
import numpy as np

#understanding vector
employee = np.array([25, 50000, 2])

print("Employee data:", employee)
```

```python
employees = np.array([[25, 50000, 2], [30, 60000, 5], [28, 55000, 3]])
print("employees \n", employees)
```

```python
age_experience = np.array([25, 2])
weights = np.array([0.5, 2])

result = np.dot(age_experience, weights)
print("Result of dot product:", result)
```

```python
A = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

result = np.dot(A, b)
print("result of dot product of A and b:\n", result)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
print("import")
```

```python
df = pd.read_csv("dataset/winequality-red.csv", sep=";")
df.head()
```

```python
# for missing values
missing_values = df.isnull().sum()
print("missing values in each column:")
print(missing_values)
```

```python
# for total missing values
total_miss = df.isnull().sum().sum()
print("total missing values", total_miss)
```

```python
# for duplicate Records
duplicate_count = df.duplicated().sum()
print("duplicated values/ rows:", duplicate_count)
```

```python
#dorping duplicates
df = df.drop_duplicates()
print("Dataset shape after removing duplicates:", df.shape)
#remaining duplicates
print("remaining duplicate rows:", df.duplicated().sum())
```

```python
# check Data Types 
print(df.dtypes)
```

```python
print("Number of numerical columns:",
      df.select_dtypes(include=np.number).shape[1])
```

```python
#Final dataset shape
print("Final number of rows:", len(df))
print("Final number of columns:", len(df.columns))
```

```python
# Descriptive Statistics
df.describe()
```

```python
#for mean only
print("Mean:")
print(df.mean(numeric_only=True))
```

```python
# for median
print("Median:")
print(df.median(numeric_only=True))
```

```python
#for standart deviation
print("standard deviation")
print(df.std(numeric_only=True))
```

```python
# Analyze wine Quality
quality_counts = df["quality"]. value_counts().sort_index()
print(quality_counts)

# for visualization
plt.Figure(figsize=(9, 5))
sns.countplot(data=df, x="quality")

plt.title("distribution of wine quality scores")
plt.xlabel("wine Quality")
plt.ylabel("Number of wines")
plt.show()
```

```python
# distribution analysis
#alcohol Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["alcohol"], bins=20, kde=True)

plt.title("distribution of alcohol content")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```

```python
#pH Distribution
plt.Figure(figsize=(8, 5))
sns.histplot(df["pH"], bins = 20, kde=True)
plt.title("Distribution of pH")
plt.xlabel("pH")
plt.ylabel("frequency")
plt.show()
```

```python
# Volatile acidity
plt.Figure(figsize= (8, 5))
sns. histplot(df["volatile acidity"], bins= 20, kde=True)
plt.title("Distribution of Volatile Acidity")
plt.xlabel("Volatile Acidity")
plt.ylabel("Frequency")

plt.show()
```

```python
# outliner analysis or box plot
plt.figure(figsize= (14, 7))
sns.boxplot(data= df)
plt.xticks(rotation = 45)
plt.title("Box plot of wine Features")
plt.show()
```

```python
# correlation Analysis
correlation = df.corr(numeric_only=True)
correlation
```

```python
# corelation with quality
quality_correlation = correlation["quality"].sort_values(ascending=False)
print(quality_correlation)

#heat map of  correlaiton
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Wine Characteristics")

plt.show()
```

```python
#alcohol vs quality
plt.Figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="quality",
    y="alcohol"
)
plt.title("alcohol content by wine quality")
plt.xlabel("wine Quality")
plt.ylabel("alcohol")
plt.show()
```

```python
# Volatile Acidity vs quality
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="quality",
    y="volatile acidity"
)

plt.title("Volatile Acidity by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Volatile Acidity")

plt.show()
```

```python
# Alcohol vs Quality Scatter Plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="alcohol",
    y="quality"
)

plt.title("Relationship Between Alcohol and Wine Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")

plt.show()
```

```python
# Group Analysis
quality_group = df.groupby("quality").mean(numeric_only=True)
quality_group

```

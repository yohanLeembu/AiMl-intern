```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)

y = x ** 2
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Plot of y = x^2")
plt.show()
```

```python
x = np.linspace(-10, 10, 100)
y = x ** 2
derivative = 2 * x
plt.plot(x, y, label="y = x^2")
plt.plot(x, derivative, label="Derivative")
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title("function and it derivative")
plt.show()
```
